```python
import pandas as pd
df = pd.read_csv('data/gst_invoices.csv')
df.head()
```

```python
class Graph:
    def __init__(self):
        self.adjacency_list = {}

    def add_node(self,gstin):
        if gstin not in self.adjacency_list:
            self.adjacency_list[gstin] = []
    
    def add_edge(self,seller_gstin,buyer_gstin,invoice_id,amount):
        self.add_node(seller_gstin)
        self.add_node(buyer_gstin)

        self.adjacency_list[seller_gstin].append(
            {
                "buyer" : buyer_gstin,
                "invoice" : invoice_id,
                "amount"  : amount
            }
        )
```

```python
g = Graph()

for _, row in df.iterrows():

    g.add_edge(
        row["seller_gstin"],
        row["buyer_gstin"],
        row["invoice_id"],
        row["amount"]
    )
```
